```python
cl_annotations1 = {
    '1': "Epithelial", '2': "Fibroblasts", '0': "T cells", '3': "Macrophages", '4': "Endothelial",
    '5': "Endothelial", '6': "smooth muscles", '7': "Epithelial", '8': "Epithelial", '9': "b cells",
    '10': "Epithelial", '11': "plasma cells", '12': "Epithelial"
}
```

```python
adata1.obs["manual_celltype_annotation"] = adata1.obs['leiden'].map(cl_annotations)


adata1.obs["manual_celltype_annotation"] = adata1.obs["manual_celltype_annotation"].fillna("unknown")


sc.pl.umap(
    adata1,
    color=["manual_celltype_annotation"],
)
```

```python
cluster_id = "0"  # Change this to the cluster of interest
cluster_cells = adata1[adata1.obs["leiden"] == cluster_id]
```

```python
import numpy as np

# Convert sparse matrix to dense format
cluster_X_dense = cluster_cells.X.toarray() if hasattr(cluster_cells.X, "toarray") else cluster_cells.X

# Find genes that have nonzero expression in at least one cell of the cluster
expressed_genes = adata1.var_names[np.any(cluster_X_dense > 0, axis=0)]

# Convert to list
expressed_genes_list = expressed_genes.tolist()
print(expressed_genes_list)
```
